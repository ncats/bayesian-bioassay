### Code for generate chemical descriptor for assays

```python
import pandas as pd
import numpy as np
import os
import glob
import rdkit
from rdkit import Chem
from rdkit.Chem import Descriptors
from rdkit.ML.Descriptors import MoleculeDescriptors
from rdkit.Chem.Draw import IPythonConsole

def generate_descriptor_value(smiles, descriptor_function):
    mol = Chem.MolFromSmiles(smiles)
    value = None
    if mol is not None:
        value = descriptor_function(mol)
    return value
```

```python
df = pd.read_csv('Outcome Matrix_updated.csv')

assay_df_all = []
for i in range(4,57):
    assay_df = df[['SAMPLE_ID', 'InChiKey', 'SAMPLE_NAME', df.columns[i], 'SMILES']].dropna(
        subset=[df.columns[i]]).reset_index()

    df_all = assay_df.copy()

    smile = df_all['SMILES']
    exactmolwt = []
    nwidx = []
    widx = []
    for j in range(0, len(df_all)):
        if smile[j] != '':
            try:
                m = Chem.MolFromSmiles(smile[j])
                tspas = Descriptors.ExactMolWt(m)
                exactmolwt.append(tspas)
                widx.append(j)
            except:
                nwidx.append(j)

    print(len(nwidx))
    print(len(widx))

    df_all2 = df_all.iloc[widx, :]

    for descriptor_tuple in Descriptors.descList:
        descriptor_name = descriptor_tuple[0]
        descriptor_function = descriptor_tuple[1]
        df_all2[descriptor_name] = df_all2['SMILES'].apply(lambda x: generate_descriptor_value(x, descriptor_function))

    assay_df_all.append(df_all2)
    print(df.columns[i])
```

```python
assay_df_all[0].columns[4]
```

```python
miss_per = []
assay_names = []
col_names = []
for j in range(len(assay_df_all)):
    print(assay_df_all[j].columns[4])
    miss_col = len(assay_df_all[j].iloc[:,6:].columns[assay_df_all[j].iloc[:,6:].isna().any()].tolist())
    for i in range(6,215):
        count = assay_df_all[j].iloc[:,i].isna().sum()
        if count > 0 and miss_col <13:
            print(assay_df_all[j].columns[i])
            a = (count/len(assay_df_all[j]))*100
            b = "{:.2f}".format(a)
            col_name = assay_df_all[j].columns[i]
            miss_per.append(b)
            col_names.append(col_name)
            assay_names.append(assay_df_all[j].columns[4])
            print('percentage', ("%.2f" % a), "%")
            print()
```
